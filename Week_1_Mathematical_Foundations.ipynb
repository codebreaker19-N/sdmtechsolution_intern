# 1. VECTORS

A vector is an ordered collection of numbers that
represents magnitude and direction.

Example:

V = [80, 90, 70]

It can represent marks of a student in three subjects.

### Vector Operations

1. Addition
2. Scalar Multiplication
3. Dot Product

Example:

A = [1, 2, 3]
B = [4, 5, 6]

A + B = [5, 7, 9]

Dot Product = (1*4) + (2*5) + (3*6) = 32

### Applications in AI/ML
- Word embeddings
- Feature representation
- Recommendation systems

```python
import numpy as np

A = np.array([1, 2, 3])
B = np.array([4, 5, 6])

print("Vector A:", A)
print("Vector B:", B)
print("Addition:", A + B)
print("Scalar Multiplication:", 2 * A)
print("Dot Product:", np.dot(A, B))
```

# 2. MATRICES

A matrix is a rectangular arrangement of numbers
in rows and columns.

Example:

A = [[1, 2],
     [3, 4]]

Shape = 2 x 2

### Matrix Operations
1. Matrix Addition
2. Matrix Multiplication
3. Transpose

### Applications in AI/ML
- Neural Networks
- Image Processing
- Deep Learning
- Dataset Representation

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("Addition:\n", A + B)
print("Multiplication:\n", np.dot(A, B))
print("Transpose:\n", A.T)
```

# 3. PROBABILITY

Probability measures how likely an event is to happen.

Formula:

P(A) = Favourable Outcomes / Total Outcomes

Example: Tossing a coin

Possible outcomes = Head, Tail

P(Head) = 1/2 = 0.5

### Important Concepts
1. Conditional Probability
2. Independent Events
3. Bayes' Theorem
4. Expected Value

### Applications in AI/ML
- Spam Detection
- Naive Bayes Classification
- Medical Diagnosis
- Prediction Systems

```python
outcomes = [1, 2, 3, 4, 5, 6]

even_numbers = [x for x in outcomes if x % 2 == 0]

probability = len(even_numbers) / len(outcomes)

print("Probability of getting an even number:", probability)
```

# 4. GRADIENT INTUITION

A gradient tells us the direction and rate of
the steepest increase of a function.

Example:

f(x) = x^2

Derivative:

f'(x) = 2x

At x = 3:

f'(3) = 6

### Gradient Descent

Gradient Descent is an optimization algorithm
used to minimize the loss function in Machine Learning.

Update Formula:

x_new = x_old - learning_rate * gradient

It moves model parameters towards a minimum
to reduce prediction errors.

### Applications in AI/ML
- Neural Network Training
- Deep Learning
- Linear Regression
- Model Optimization

```python
x = 4
learning_rate = 0.1

for i in range(10):
    gradient = 2 * x
    x = x - learning_rate * gradient

    print("Iteration:", i + 1, "Value:", x)

print("Final Value:", x)
```
