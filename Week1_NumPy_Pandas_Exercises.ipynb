```python

# WEEK 1: NUMPY AND PANDAS - 20 EXERCISES

import numpy as np
import pandas as pd

print("=" * 60)
print("             NUMPY EXERCISES (1-10)")
print("=" * 60)

# Q1. Create a 1D array from 1 to 10
arr = np.arange(1, 11)
print("\nQ1. Array from 1 to 10:", arr)

# Q2. Create a 3x3 array of zeros
arr = np.zeros((3, 3))
print("\nQ2. 3x3 Zero Matrix:\n", arr)

# Q3. Create a 3x3 identity matrix
arr = np.eye(3)
print("\nQ3. Identity Matrix:\n", arr)

# Q4. Create even numbers between 1 and 20
arr = np.arange(2, 21, 2)
print("\nQ4. Even Numbers:", arr)

# Q5. Find mean, median and standard deviation
arr = np.array([10, 20, 30, 40, 50])
print("\nQ5. Mean:", np.mean(arr))
print("Median:", np.median(arr))
print("Standard Deviation:", np.std(arr))

# Q6. Find maximum and minimum
arr = np.array([45, 12, 78, 23, 9, 56])
print("\nQ6. Maximum:", np.max(arr))
print("Minimum:", np.min(arr))

# Q7. Reshape array into 3x3 matrix
arr = np.arange(1, 10)
matrix = arr.reshape(3, 3)
print("\nQ7. Reshaped Matrix:\n", matrix)

# Q8. Element-wise addition and multiplication
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print("\nQ8. Addition:", a + b)
print("Multiplication:", a * b)

# Q9. Extract elements greater than 5
arr = np.array([2, 5, 8, 1, 9, 4, 7])
print("\nQ9. Elements greater than 5:", arr[arr > 5])

# Q10. Generate 10 random integers between 1 and 100
arr = np.random.randint(1, 101, size=10)
print("\nQ10. Random Numbers:", arr)


print("\n" + "=" * 60)
print("             PANDAS EXERCISES (11-20)")
print("=" * 60)

# Import Pandas
print("\nPandas Version:", pd.__version__)

# Create DataFrame
data = {
    "Name": ["Aman", "Nidhi", "Rahul", "Priya", "Karan"],
    "Age": [21, 20, 22, 19, 23],
    "Marks": [85, 92, 76, 95, 68],
    "City": ["Delhi", "Delhi", "Mumbai", "Pune", "Delhi"]
}

df = pd.DataFrame(data)

# Q11. Display first three rows
print("\nQ11. First Three Rows:\n", df.head(3))

# Q12. Display last two rows
print("\nQ12. Last Two Rows:\n", df.tail(2))

# Q13. Display shape of DataFrame
print("\nQ13. DataFrame Shape:", df.shape)

# Q14. Display column names and data types
print("\nQ14. Column Names:", df.columns.tolist())
print("Data Types:\n", df.dtypes)

# Q15. Find average marks
print("\nQ15. Average Marks:", df["Marks"].mean())

# Q16. Find student with highest marks
print("\nQ16. Student with Highest Marks:\n",
      df.loc[df["Marks"].idxmax()])

# Q17. Filter students with marks greater than 80
print("\nQ17. Students with Marks > 80:\n",
      df[df["Marks"] > 80])

# Q18. Sort students by marks in descending order
print("\nQ18. Sorted DataFrame:\n",
      df.sort_values(by="Marks", ascending=False))

# Q19. Count students in each city
print("\nQ19. Students in Each City:\n",
      df["City"].value_counts())

# Q20. Add Result column
df["Result"] = df["Marks"].apply(
    lambda x: "Pass" if x >= 40 else "Fail"
)
print("\nQ20. DataFrame with Result Column:\n", df)

print("\n" + "=" * 60)
print("       ALL 20 EXERCISES COMPLETED SUCCESSFULLY!")
print("=" * 60)
```
